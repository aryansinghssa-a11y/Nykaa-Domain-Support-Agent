## 2. Create the Markdown files

Run this cell to write each policy into its own `.md` file.

```python
for filename, content in POLICY_DOCUMENTS.items():
    (KNOWLEDGE_BASE_DIR / filename).write_text(content.strip() + '\n', encoding='utf-8')

print(f'Created {len(POLICY_DOCUMENTS)} policy documents.')
for filename in sorted(POLICY_DOCUMENTS):
    print('-', filename)
```

## 3. Validate topic coverage and sentence counts

The assignment requires at least 12 documents, every required topic, and 2–5 sentences per document.

```python
REQUIRED_FILES = [
    'return_window.md',
    'cod_refund.md',
    'delivery_sla.md',
    'reverse_pickup.md',
    'warranty.md',
    'cancellation.md',
    'loyalty_points.md',
    'payment_failure.md',
    'size_exchange.md',
    'damaged_item.md',
    'international_shipping.md',
    'escalation_matrix.md',
]

def count_sentences(text):
    return len(re.findall(r'[.!?](?=\s|$)', text.strip()))
errors = []
print('KNOWLEDGE BASE VALIDATION')
print('=' * 60)

for filename in REQUIRED_FILES:
    path = KNOWLEDGE_BASE_DIR / filename

    if not path.exists():
        errors.append(f'Missing required file: {filename}')
        print(f'FAIL | {filename} | missing')
        continue

    content = path.read_text(encoding='utf-8').strip()
    sentence_count = count_sentences(content)

    passed = bool(content) and 2 <= sentence_count <= 5
    print(f"{'PASS' if passed else 'FAIL'} | {filename:25s} | {sentence_count} sentences")

    if not passed:
        errors.append(f'{filename} has an invalid sentence count or is empty.')

if len(list(KNOWLEDGE_BASE_DIR.glob('*.md'))) < 12:
    errors.append('Fewer than 12 Markdown documents exist.')

if errors:
    print('\\nVALIDATION ERRORS:')
    for error in errors:
        print(' -', error)
    raise ValueError('Knowledge-base validation failed.')

print('\\nPASS: All 12 required topics are present.')
print('PASS: Every document contains 2–5 sentences.')
print('PASS: Task 2 knowledge-base validation completed.')
```
